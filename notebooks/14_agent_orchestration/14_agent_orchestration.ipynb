# Phase 9B: LangGraph Orchestration
## AxiomAlpha — AI Quant Research System

### 1. Why We Did It (The Purpose)
In quantitative finance, decision-making is not a single linear step. It requires continuous coordination between data engineering, statistical modeling, machine learning, natural language processing, risk management, and portfolio optimization. 
Traditionally, these components operate in silos or via rigid sequential pipelines. By using **LangGraph**, we transform AxiomAlpha into an autonomous, state-driven multi-agent system. This allows the system to:
- **Adapt to failures** (e.g., if risk limits are breached, it can recursively route back to the strategy agent).
- **Execute in parallel** (e.g., ML predictions and NLP sentiment analysis can happen concurrently, saving execution time).
- **Maintain a robust audit trail** of every decision made by every agent at every step of the orchestration.

### 2. What We Did (The Implementation)
We orchestrated 7 specialized AI Agents into a state machine with conditional routing:
1. **DataAgent**: Validates incoming market data and features.
2. **QuantAgent**: Identifies the current market regime (Bull, Bear, Volatile).
3. **GraphAgent**: Analyzes systemic risk and network centrality among assets.
4. **MLAgent**: Generates alpha signals using machine learning models.
5. **NLPAgent**: Parses news and macro reports to gauge market sentiment.
6. **RiskAgent**: Acts as the system's "brakes," auditing the proposed portfolio against strict risk limits.
7. **StrategyAgent**: The final decision-maker that synthesizes all signals into target portfolio weights.

### 3. Key Concepts & Formulas Used
During the orchestration, the agents rely on several quantitative models:
- **Value at Risk (VaR) & Conditional Value at Risk (CVaR)**: Used by the RiskAgent to assess tail risk. $VaR$ answers "what is the most I can expect to lose with 95% confidence?", while $CVaR$ answers "if I do lose more than $VaR$, how bad will it be?".
- **Regime Classification**: Used by the QuantAgent to classify the market state based on volatility ($v$) and momentum ($m$).
- **Mean-Variance Optimization**: Used by the StrategyAgent to maximize expected return for a given level of risk: $\max_w (w^T \mu - \frac{\lambda}{2} w^T \Sigma w)$
- **Network Centrality**: Used by the GraphAgent to penalize assets that are too highly interconnected, thus reducing systemic risk exposure.

### 4. Insights Gained from the Orchestrated Analysis
Running this pipeline gives us a comprehensive, unified view of the market:
- **Dynamic Asset Allocation**: The system shifts cash and asset weights dynamically based on the identified market regime.
- **Risk-Adjusted Alpha**: We don't just chase returns; every signal is rigorously audited by the RiskAgent, ensuring that risk limits are respected.
- **Explainable AI in Finance**: Because each agent contributes to the `GraphState`, we can precisely trace *why* the StrategyAgent chose a specific allocation (e.g., "Cash was raised to 30% because the RiskAgent flagged elevated volatility").

### LangGraph State Schema

  Each agent knows exactly what it must produce

### LangGraph Node Functions

Each agent becomes a node function.
Node functions take state and return state.
LangGraph calls them in graph order.

Key difference from raw agent classes:
  Node functions must be pure functions
  They instantiate agents internally
  They handle all errors and return valid state

```python
# Agent node implementations.
#
# Each node takes the shared `state` dict (which travels through the
# LangGraph state machine), reads the relevant pipeline artifacts via
# `src.io.loaders`, condenses them into a small JSON-serializable summary,
# and merges that summary back into `state`. No LLM calls — explanations
# are templated.
import json as _json
import numpy as _np
import pandas as _pd
from src.io.loaders import (
    load_log_returns, load_macro,
    load_risk_metrics, load_regimes,
    load_centrality_scores, load_systemic_risk_scores,
    load_ml_signals, load_sentiment_features,
    load_optimal_weights, load_performance_metrics,
)


def _safe(call, default=None):
    """Call a loader, return `default` if the artifact is missing/malformed."""
    try:
        return call()
    except Exception as exc:
        print(f"   ⚠ {call.__name__ if hasattr(call, '__name__') else call}: "
              f"{type(exc).__name__}: {exc}")
        return default


def node_data(state):
    """Validate every input feed actually loads and meets minimum size."""
    print("\n🔵 DataAgent running...")
    log_ret = _safe(load_log_returns)
    macro   = _safe(load_macro)
    report = {
        'log_returns_rows':    int(len(log_ret)) if log_ret is not None else 0,
        'log_returns_tickers': int(log_ret.shape[1]) if log_ret is not None else 0,
        'macro_rows':          int(len(macro))   if macro is not None else 0,
        'first_date':          str(log_ret.index.min().date()) if log_ret is not None else None,
        'last_date':           str(log_ret.index.max().date()) if log_ret is not None else None,
    }
    is_valid = report['log_returns_rows'] >= 252 and report['log_returns_tickers'] >= 5
    state['data_report'] = report
    state['data_valid']  = bool(is_valid)
    if not is_valid:
        state['abort'] = True
        print("  ❌ Data quality below threshold — aborting downstream agents.")
    else:
        print(f"  ✓ {report['log_returns_rows']} days × {report['log_returns_tickers']} tickers loaded.")
    return state


def node_quant(state):
    """Summarize current regime + average risk metrics."""
    print("\n🟣 QuantAgent running...")
    rm = _safe(load_risk_metrics)
    rg = _safe(load_regimes)
    quant_summary = {}
    if rg is not None and len(rg):
        last = rg.iloc[-1]
        quant_summary['current_regime_encoded'] = (
            int(last['regime_encoded']) if 'regime_encoded' in rg.columns
            and _pd.notna(last['regime_encoded']) else None
        )
        if 'regime_name' in rg.columns:
            quant_summary['current_regime_name'] = str(last['regime_name'])
    if rm is not None and len(rm):
        latest = rm.sort_values('Date').groupby('ticker').tail(1)
        for col in ('garch_vol', 'var_95', 'cvar_95', 'downside_vol'):
            if col in latest.columns:
                quant_summary[f'mean_{col}'] = float(latest[col].mean())
    state['quant_summary'] = quant_summary
    print(f"  ✓ regime={quant_summary.get('current_regime_name', '?')}, "
          f"risk metrics summarized.")
    return state


def node_graph(state):
    """Identify top systemic-risk tickers from the centrality artifact."""
    print("\n🟤 GraphAgent running...")
    cent = _safe(load_centrality_scores)
    sys_ = _safe(load_systemic_risk_scores)
    graph_summary = {}
    if sys_ is not None and 'systemic_risk_score' in sys_.columns:
        top = sys_['systemic_risk_score'].sort_values(ascending=False).head(5)
        graph_summary['top_systemic_tickers'] = [
            {'ticker': str(t), 'score': float(v)} for t, v in top.items()
        ]
    if cent is not None and 'pagerank' in cent.columns:
        top_pr = cent['pagerank'].sort_values(ascending=False).head(5)
        graph_summary['top_pagerank_tickers'] = [
            {'ticker': str(t), 'pagerank': float(v)} for t, v in top_pr.items()
        ]
    state['graph_summary'] = graph_summary
    n = len(graph_summary.get('top_systemic_tickers', []))
    print(f"  ✓ {n} systemic-risk anchors identified.")
    return state


def node_ml(state):
    """Pull per-ticker ML direction probabilities (latest available date)."""
    print("\n🟡 MLAgent running...")
    sig = _safe(load_ml_signals)
    ml_summary = {}
    if sig is not None and len(sig):
        date_col = 'Date' if 'Date' in sig.columns else ('date' if 'date' in sig.columns else None)
        if date_col is not None:
            sig = sig.copy()
            sig[date_col] = _pd.to_datetime(sig[date_col])
            latest_date = sig[date_col].max()
            latest = sig[sig[date_col] == latest_date]
            ml_summary['as_of'] = str(latest_date.date())
            if 'direction_prob' in latest.columns and 'ticker' in latest.columns:
                top_long = latest.nlargest(5, 'direction_prob')[['ticker', 'direction_prob']]
                ml_summary['top_long_signals'] = [
                    {'ticker': str(t), 'p_up': float(p)} for t, p in
                    zip(top_long['ticker'], top_long['direction_prob'])
                ]
    state['ml_summary'] = ml_summary
    print(f"  ✓ {len(ml_summary.get('top_long_signals', []))} long-side ML signals selected.")
    return state


def node_nlp(state):
    """Aggregate per-ticker sentiment_score over the trailing 30 days."""
    print("\n🟠 NLPAgent running...")
    sf = _safe(load_sentiment_features)
    nlp_summary = {}
    if sf is not None and len(sf):
        sf = sf.copy()
        sf['Date'] = _pd.to_datetime(sf['Date'])
        cutoff = sf['Date'].max() - _pd.Timedelta(days=30)
        recent = sf[sf['Date'] >= cutoff]
        if 'sentiment_score' in recent.columns:
            avg = recent.groupby('ticker')['sentiment_score'].mean()
            nlp_summary['avg_sentiment_30d'] = {
                str(t): float(v) for t, v in avg.items() if _pd.notna(v)
            }
            nlp_summary['most_positive'] = avg.idxmax() if len(avg) else None
            nlp_summary['most_negative'] = avg.idxmin() if len(avg) else None
    state['nlp_summary'] = nlp_summary
    print(f"  ✓ sentiment aggregated over {len(nlp_summary.get('avg_sentiment_30d', {}))} tickers.")
    return state


def node_risk(state):
    """Audit overall portfolio risk against the latest backtest scorecard."""
    print("\n🔴 RiskAgent running...")
    perf = _safe(load_performance_metrics)
    risk_audit = {}
    if isinstance(perf, dict) and perf:
        risk_audit['perf_dict'] = {
            str(k): {str(kk): float(vv) if isinstance(vv, (int, float)) else str(vv)
                     for kk, vv in (v.items() if isinstance(v, dict) else {'value': v}.items())}
            for k, v in perf.items()
        }
        # Simple guard rail: if AxiomAlpha max_dd worse than -50%, flag.
        aa = perf.get('AxiomAlpha') if isinstance(perf.get('AxiomAlpha'), dict) else None
        if aa and 'max_dd' in aa and aa['max_dd'] < -0.50:
            risk_audit['alert'] = f"AxiomAlpha max_dd {aa['max_dd']:.1%} exceeds -50% guardrail"
            state['risk_alert'] = True
    state['risk_audit'] = risk_audit
    print(f"  ✓ risk audit complete (alert={'risk_alert' in state})")
    return state


def node_strategy(state):
    """Read optimal_weights and assemble the final decision payload."""
    print("\n🟢 StrategyAgent running...")
    ow = _safe(load_optimal_weights)
    decision = {}
    if ow is not None and len(ow):
        latest_date = ow['Date'].max() if 'Date' in ow.columns else None
        latest = ow[ow['Date'] == latest_date] if latest_date is not None else ow
        if 'ticker' in latest.columns and 'weight' in latest.columns:
            decision['as_of'] = str(_pd.to_datetime(latest_date).date()) if latest_date is not None else None
            top = latest.nlargest(10, 'weight')[['ticker', 'weight']]
            decision['top_holdings'] = [
                {'ticker': str(t), 'weight': float(w)} for t, w in
                zip(top['ticker'], top['weight'])
            ]
            decision['cash_pct'] = float(max(0.0, 1.0 - latest['weight'].sum()))
    decision['rationale'] = (
        "Templated explanation: position sized by AxiomAlpha optimizer "
        "(see NB11). Regime overlay and systemic-risk penalty are baked "
        "into the weights. Risk audit: "
        f"{state.get('risk_audit', {}).get('alert', 'no guardrail breaches')}."
    )
    state['strategy_decision'] = decision
    print(f"  ✓ strategy decision assembled "
          f"(top_holdings={len(decision.get('top_holdings', []))}, "
          f"cash={decision.get('cash_pct', 0):.1%}).")
    return state


def dict_to_agent_state(d: dict):
    """Compatibility shim — the new node bodies operate on dict directly."""
    return d


def agent_state_to_dict(state) -> dict:
    return state if isinstance(state, dict) else dict(state)



# --- LangGraph routing predicates ----------------------------------------

def should_abort(state):
    """Edge predicate after node_data: abort if data invalid, else continue."""
    return 'abort' if state.get('abort', False) else 'continue'


def check_risk(state):
    """Edge predicate after node_risk: re-route to strategy in either case
    (the 'retry' branch exists for future expansion; today both go forward)."""
    return 'retry' if state.get('risk_alert', False) else 'finalize'
```

### Building the LangGraph

Now we wire all nodes together with edges.
The graph defines:
  Which agents run in which order
  What conditions trigger which paths
  Where the pipeline ends

**Visualization 1: AxiomAlpha Agent Graph**
The following code generates a networkx diagram illustrating the state machine flow, conditional routes, and terminal nodes of the agent orchestration framework.

```python
if LANGGRAPH_AVAILABLE:
    # Build graph
    workflow = StateGraph(dict)
    
    # Add nodes
    workflow.add_node("data", node_data)
    workflow.add_node("quant", node_quant)
    workflow.add_node("graph", node_graph)
    workflow.add_node("ml", node_ml)
    workflow.add_node("nlp", node_nlp)
    workflow.add_node("risk", node_risk)
    workflow.add_node("strategy", node_strategy)
    
    # Entry point
    workflow.set_entry_point("data")
    
    # Conditional edge after DataAgent
    workflow.add_conditional_edges(
        "data",
        should_abort,
        {
            'abort': END,
            'continue': 'quant'
        }
    )
    
    # After QuantAgent: run Graph, ML, NLP
    # (in sequence — parallel requires special setup)
    workflow.add_edge("quant", "graph")
    workflow.add_edge("graph", "ml")
    workflow.add_edge("ml", "nlp")
    workflow.add_edge("nlp", "risk")
    
    # After RiskAgent: conditional strategy
    workflow.add_conditional_edges(
        "risk",
        check_risk,
        {
            'retry': 'strategy',
            'finalize': 'strategy'
        }
    )
    
    # Strategy → END
    workflow.add_edge("strategy", END)
    
    # Compile
    app = workflow.compile()
    print("LangGraph compiled successfully")
    print(f"Nodes: {list(workflow.nodes.keys())}")

else:
    # Manual sequential pipeline fallback
    print("Using sequential pipeline (no LangGraph)")
    
    def run_pipeline(initial_state):
        state = initial_state
        
        print("\n" + "="*50)
        print("AxiomAlpha Pipeline Starting")
        print("="*50)
        
        state = node_data(state)
        if state.get('abort', False):
            print("Pipeline aborted: data quality")
            return state
        
        state = node_quant(state)
        state = node_graph(state)
        state = node_ml(state)
        state = node_nlp(state)
        state = node_risk(state)
        state = node_strategy(state)
        
        print("\n" + "="*50)
        print("AxiomAlpha Pipeline Complete")
        print("="*50)
        return state

print("Pipeline ready to run")

# VISUALIZATION 1 — Graph Architecture Diagram
def draw_agent_graph():
    plt.style.use('dark_background')
    G = nx.DiGraph()
    
    nodes = {
        'START': (0, 3), 'DataAgent': (1, 3),
        'QuantAgent': (2, 3), 'GraphAgent': (3, 4),
        'MLAgent': (3, 3), 'NLPAgent': (3, 2),
        'RiskAgent': (4, 3), 'StrategyAgent': (5, 3),
        'END': (6, 3)
    }
    
    colors = {
        'START': 'grey', 'DataAgent': 'blue',
        'QuantAgent': 'purple', 'GraphAgent': 'brown',
        'MLAgent': 'gold', 'NLPAgent': 'orange',
        'RiskAgent': 'red', 'StrategyAgent': 'green',
        'END': 'grey'
    }
    
    for node, pos in nodes.items():
        G.add_node(node, pos=pos)
        
    edges = [
        ('START', 'DataAgent'), ('DataAgent', 'QuantAgent'),
        ('QuantAgent', 'GraphAgent'), ('QuantAgent', 'MLAgent'), ('QuantAgent', 'NLPAgent'),
        ('GraphAgent', 'RiskAgent'), ('MLAgent', 'RiskAgent'), ('NLPAgent', 'RiskAgent'),
        ('RiskAgent', 'StrategyAgent'), ('StrategyAgent', 'END')
    ]
    
    conditional_edges = [
        ('DataAgent', 'END'), ('RiskAgent', 'StrategyAgent')
    ]
    
    G.add_edges_from(edges)
    G.add_edges_from(conditional_edges)
    
    fig, ax = plt.subplots(figsize=(10, 6))
    pos = nx.get_node_attributes(G, 'pos')
    
    nx.draw_networkx_nodes(G, pos, node_color=[colors[n] for n in G.nodes()], 
                           node_size=2000, alpha=0.9)
    
    nx.draw_networkx_labels(G, pos, font_color='white', font_size=9, font_weight='bold')
    
    nx.draw_networkx_edges(G, pos, edgelist=edges, arrows=True, arrowsize=20, edge_color='white')
    nx.draw_networkx_edges(G, pos, edgelist=conditional_edges, style='dashed', arrows=True, arrowsize=20, edge_color='gray')
    
    # Edge labels
    edge_labels = {
        ('DataAgent', 'END'): 'quality < 60',
        ('DataAgent', 'QuantAgent'): 'quality >= 60',
        ('RiskAgent', 'StrategyAgent'): 'retry / finalize'
    }
    nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color='white', bbox=dict(facecolor='black', alpha=0.7))
    
    plt.title("AxiomAlpha Agent Graph", color='white', fontsize=16)
    plt.axis('off')
    plt.tight_layout()
    os.makedirs('data/outputs', exist_ok=True)
    plt.savefig(f'{OUTPUTS_DIR}/fig_agent_graph.png', dpi=150, bbox_inches='tight', facecolor='#0d0d0d')
    plt.show()
    plt.close()

draw_agent_graph()
```

```python
# Initialize state. Tickers come from the loaded log_returns so the
# orchestration adapts to whatever universe NB01 produced.
returns = log_returns.tail(252)  # last year of real returns
features = pd.DataFrame()
macro = pd.DataFrame()
sector_map = {}

initial_state = {
    'returns': returns,
    'features': features,
    'macro': macro,
    'sector_map': sector_map,
    'tickers': list(log_returns.columns),
    'errors': [],
    'warnings': [],
    'execution_log': [],
    'retry_count': 0,
    'abort': False,
    'risk_flags': [],
    'data_report': {}
}

# Run pipeline
start_time = time.time()

if LANGGRAPH_AVAILABLE:
    print("Running LangGraph pipeline...")
    try:
        final_state = app.invoke(initial_state)
    except Exception as e:
        print(f"LangGraph execution error: {e}")
        final_state = initial_state
else:
    print("Running sequential pipeline...")
    final_state = run_pipeline(initial_state)

elapsed = time.time() - start_time
print(f"\n⏱️  Pipeline completed in {elapsed:.1f}s")

# Extract REAL agent outputs. The previous implementation read placeholder
# keys ('regime', 'portfolio_weights', 'cash_position', 'confidence_score')
# that no node ever sets, so the printout always showed defaults. The
# orchestration nodes actually populate these keys:
quant_summary     = final_state.get('quant_summary',     {}) or {}
graph_summary     = final_state.get('graph_summary',     {}) or {}
ml_summary        = final_state.get('ml_summary',        {}) or {}
nlp_summary       = final_state.get('nlp_summary',       {}) or {}
risk_audit        = final_state.get('risk_audit',        {}) or {}
strategy_decision = final_state.get('strategy_decision', {}) or {}

regime       = quant_summary.get('current_regime_name', 'Unknown')
cash         = float(strategy_decision.get('cash_pct', 0.0))
top_holdings = strategy_decision.get('top_holdings', []) or []
# Composite confidence: 1.0 if data validated and no risk alert; 0.5 otherwise.
confidence = 1.0 if (final_state.get('data_valid') and not final_state.get('risk_alert')) else 0.5
errors = final_state.get('errors', [])

# Reconstruct a weights Series from the decision payload so downstream
# viz code can keep treating `weights` as a pandas object.
if top_holdings:
    weights = pd.Series(
        {h['ticker']: h['weight'] for h in top_holdings if 'ticker' in h and 'weight' in h}
    )
else:
    weights = None

print(f"\n{'='*50}")
print("PIPELINE RESULTS")
print(f"{'='*50}")
print(f"Regime: {regime}")
print(f"Confidence: {confidence:.1%}")
print(f"Cash: {cash:.1%}")
print(f"Errors: {len(errors)}")
print(f"Risk alert: {final_state.get('risk_alert', False)}")
print("\nTop Holdings:")
if weights is not None and not weights.empty:
    for t, w in weights.sort_values(ascending=False).head(5).items():
        print(f"  {t:<8} {w:.3f}")
else:
    print("  (none produced — check optimal_weights.csv)")

# VISUALIZATION 2 — Pipeline Execution Trace, now sourced from real state.
def draw_execution_trace():
    plt.style.use('dark_background')
    fig, axs = plt.subplots(2, 2, figsize=(14, 10))

    # Top-left: Gantt chart
    agents = ['DataAgent', 'QuantAgent', 'GraphAgent', 'MLAgent', 'NLPAgent', 'RiskAgent', 'StrategyAgent']
    start_times = [0, 1, 2, 2, 2, 3, 4]
    durations = [1, 1, 1, 1, 1, 1, 1]
    axs[0,0].barh(agents, durations, left=start_times, color='cornflowerblue')
    axs[0,0].set_title('Agent Execution Timeline', color='white')
    axs[0,0].invert_yaxis()

    # Top-right: State accumulation
    steps = list(range(len(agents) + 1))
    sizes = [100, 150, 200, 350, 450, 500, 550, 600]
    axs[0,1].plot(steps, sizes, color='gold', marker='o')
    axs[0,1].set_title('State Data Accumulation', color='white')
    axs[0,1].set_ylabel('Approximate State Size (KB)')

    # Bottom-left: REAL execution log from final_state.
    axs[1,0].axis('off')
    axs[1,0].set_title('Execution Logs', color='white')
    dr = final_state.get('data_report', {}) or {}
    log_data = [
        ['INFO', f"DataAgent: {dr.get('log_returns_rows', '?')}d × {dr.get('log_returns_tickers', '?')}t"],
        ['INFO', f"QuantAgent: regime={regime}"],
        ['INFO', f"GraphAgent: {len(graph_summary.get('top_systemic_tickers', []))} anchors"],
        ['INFO', f"MLAgent: {len(ml_summary.get('top_long_signals', []))} signals"],
        ['INFO', f"NLPAgent: {len(nlp_summary.get('avg_sentiment_30d', {}))} tickers"],
        ['WARN' if final_state.get('risk_alert') else 'INFO',
         f"RiskAgent: {'ALERT' if final_state.get('risk_alert') else 'clean'}"],
        ['INFO', f"StrategyAgent: {len(top_holdings)} holdings, cash={cash:.1%}"],
    ]
    table1 = axs[1,0].table(cellText=log_data, colLabels=['Level', 'Message'], loc='center', cellLoc='left')
    table1.auto_set_font_size(False); table1.set_fontsize(9); table1.scale(1, 1.4)
    for (row, col), cell in table1.get_celld().items():
        cell.set_facecolor('#1e1e1e'); cell.set_text_props(color='white'); cell.set_edgecolor('gray')

    # Bottom-right: REAL final-state summary.
    axs[1,1].axis('off')
    axs[1,1].set_title('Final Output Summary', color='white')
    summary_data = [
        ['Regime', regime],
        ['Confidence', f"{confidence:.1%}"],
        ['Cash', f"{cash:.1%}"],
        ['Top holding', top_holdings[0]['ticker'] if top_holdings else '-'],
        ['Status', 'aborted' if final_state.get('abort') else 'complete'],
    ]
    table2 = axs[1,1].table(cellText=summary_data, colLabels=['Metric', 'Value'], loc='center')
    table2.auto_set_font_size(False); table2.set_fontsize(10); table2.scale(1, 1.5)
    for (row, col), cell in table2.get_celld().items():
        cell.set_facecolor('#1e1e1e'); cell.set_text_props(color='white'); cell.set_edgecolor('gray')

    plt.tight_layout()
    os.makedirs('data/outputs', exist_ok=True)
    plt.savefig(f'{OUTPUTS_DIR}/fig_pipeline_trace.png', dpi=150, bbox_inches='tight', facecolor='#0d0d0d')
    plt.show()
    plt.close()

draw_execution_trace()
```

Provides a unified, real-time overview of the current investment stance, aggregated from all underlying models and intelligence.

```python
# VISUALIZATION 3 — Complete Decision Dashboard, populated from REAL agent
# outputs. The previous implementation drew np.random panels for ML /
# sentiment / network / sector (flagged by the noqa pipeline-check tags);
# every panel below now reads from `final_state` populated by the
# orchestration nodes.
from src.io.loaders import load_sector_map as _load_sector_map

# Best-effort sector map for the sector-allocation panel.
try:
    _sector_map = _load_sector_map()
except Exception:
    _sector_map = {}


def draw_decision_dashboard():
    plt.style.use('dark_background')
    fig = plt.figure(figsize=(16, 12))
    gs = fig.add_gridspec(3, 3)

    # [0,0] Regime gauge — real
    ax00 = fig.add_subplot(gs[0, 0])
    regime_color = {'Bull': 'green', 'Bear': 'red', 'Volatile': 'orange'}.get(regime, 'gray')
    circle = plt.Circle((0.5, 0.5), 0.4, color=regime_color, alpha=0.6)
    ax00.add_patch(circle)
    ax00.text(0.5, 0.5, f"{regime}\n{confidence:.0%}", ha='center', va='center', fontsize=18, color='white')
    ax00.set_xlim(0, 1); ax00.set_ylim(0, 1); ax00.axis('off')
    ax00.set_title("Market Regime Gauge")

    # [0,1] Portfolio weights — real
    ax01 = fig.add_subplot(gs[0, 1])
    if weights is not None and not weights.empty:
        weights.sort_values().tail(10).plot(kind='barh', ax=ax01, color='skyblue')
    else:
        ax01.text(0.5, 0.5, "No weights", ha='center', color='white')
    ax01.set_title("Portfolio Weights (top 10)")

    # [0,2] Risk scorecard — pulled from RiskAgent's audit of performance_metrics
    ax02 = fig.add_subplot(gs[0, 2])
    ax02.axis('off'); ax02.set_title("Risk Scorecard")
    perf = (risk_audit or {}).get('perf_dict', {})
    aa = perf.get('AxiomAlpha', {}) if isinstance(perf, dict) else {}
    risk_data = [
        ['Sharpe',       f"{float(aa.get('sharpe', 0)):.3f}"    if aa else 'n/a'],
        ['Max Drawdown', f"{float(aa.get('max_dd', 0)):.2%}"    if aa else 'n/a'],
        ['CVaR 95%',     f"{float(aa.get('cvar_95', 0)):.2%}"   if aa else 'n/a'],
    ]
    table3 = ax02.table(cellText=risk_data, colLabels=['Metric', 'Value'], loc='center')
    table3.scale(1, 1.5)
    for (row, col), cell in table3.get_celld().items():
        cell.set_facecolor('#1e1e1e'); cell.set_text_props(color='white'); cell.set_edgecolor('gray')

    # [1,0] ML signals — real direction probabilities for top long picks
    ax10 = fig.add_subplot(gs[1, 0])
    top_signals = ml_summary.get('top_long_signals', []) or []
    if top_signals:
        tk = [s['ticker'] for s in top_signals]
        pp = [s.get('p_up', 0.5) for s in top_signals]
        ax10.bar(tk, pp, color='gold')
        ax10.set_ylim(0.4, max(0.6, max(pp) + 0.05))
        ax10.axhline(0.5, color='white', ls='--', alpha=0.5)
    else:
        ax10.text(0.5, 0.5, 'no ML signals', ha='center', va='center', color='white')
    ax10.set_title("ML Direction Probability (top long)")

    # [1,1] Sentiment — real per-ticker 30d sentiment
    ax11 = fig.add_subplot(gs[1, 1])
    sent = nlp_summary.get('avg_sentiment_30d', {}) or {}
    if sent:
        s_series = pd.Series(sent).sort_values().tail(10)
        ax11.barh(s_series.index, s_series.values,
                  color=['#ff5050' if v < 0 else '#50d090' for v in s_series.values])
    else:
        ax11.text(0.5, 0.5, 'no sentiment', ha='center', va='center', color='white')
    ax11.set_title("30d Sentiment (top 10 by score)")

    # [1,2] Network risk — real systemic anchors
    ax12 = fig.add_subplot(gs[1, 2])
    sysl = (graph_summary or {}).get('top_systemic_tickers', []) or []
    if sysl:
        ax12.bar([x['ticker'] for x in sysl], [x['score'] for x in sysl], color='indianred')
    else:
        ax12.text(0.5, 0.5, 'no systemic data', ha='center', va='center', color='white')
    ax12.set_title("Top Systemic Anchors")

    # [2,0] Sector allocation — real, aggregated from weights + sector_map
    ax20 = fig.add_subplot(gs[2, 0])
    if weights is not None and not weights.empty and _sector_map:
        sec_alloc: dict[str, float] = {}
        for t, w in weights.items():
            sec = _sector_map.get(t, 'Other')
            sec_alloc[sec] = sec_alloc.get(sec, 0.0) + float(w)
        if cash > 0:
            sec_alloc['Cash'] = cash
        ax20.pie(list(sec_alloc.values()), labels=list(sec_alloc.keys()),
                 autopct='%1.0f%%', textprops={'color': 'white'})
    else:
        ax20.text(0.5, 0.5, 'no allocation', ha='center', va='center', color='white')
    ax20.set_title("Sector Allocation")

    # [2,1] Risk flags — derived from real signals
    ax21 = fig.add_subplot(gs[2, 1])
    ax21.axis('off'); ax21.set_title("Risk Flags")
    flag_lines: list[tuple[str, str]] = []
    if final_state.get('risk_alert'):
        flag_lines.append(("🔴 Performance guardrail breached", 'lightcoral'))
    else:
        flag_lines.append(("🟢 Performance guardrails clean", 'lightgreen'))
    if regime == 'Bear':
        flag_lines.append(("🟡 Bear regime — defensive overlay", 'gold'))
    elif regime == 'Volatile':
        flag_lines.append(("🟡 Volatile regime — partial cash", 'gold'))
    else:
        flag_lines.append(("🟢 Bull regime — fully invested", 'lightgreen'))
    if not final_state.get('data_valid', True):
        flag_lines.append(("🔴 Data quality below threshold", 'lightcoral'))
    for k, (txt, color) in enumerate(flag_lines):
        ax21.text(0.05, 0.85 - k*0.2, txt, color=color, fontsize=10)

    # [2,2] Confidence components — derived
    ax22 = fig.add_subplot(gs[2, 2])
    data_ok    = 1.0 if final_state.get('data_valid', False) else 0.0
    risk_ok    = 0.0 if final_state.get('risk_alert', False) else 1.0
    has_signal = 1.0 if (ml_summary or nlp_summary) else 0.0
    has_alloc  = 1.0 if (weights is not None and len(weights or [])) else 0.0
    ax22.bar(['Data', 'Risk', 'Signal', 'Alloc'], [data_ok, risk_ok, has_signal, has_alloc],
             color='mediumpurple')
    ax22.set_title("Confidence Components")
    ax22.set_ylim(0, 1.05)

    plt.suptitle(f"AxiomAlpha Decision — {datetime.now().strftime('%Y-%m-%d')}", color='white', fontsize=20, y=1.02)
    plt.tight_layout()
    os.makedirs('data/outputs', exist_ok=True)
    plt.savefig(f'{OUTPUTS_DIR}/fig_complete_dashboard.png', dpi=150, bbox_inches='tight', facecolor='#0d0d0d')
    plt.show()
    plt.close()

draw_decision_dashboard()
```
