# Phase 9B: LangGraph Orchestration
## AxiomAlpha — AI Quant Research System

### 1. Why We Did It (The Purpose)
In quantitative finance, decision-making is not a single linear step. It requires continuous coordination between data engineering, statistical modeling, machine learning, natural language processing, risk management, and portfolio optimization. 
Traditionally, these components operate in silos or via rigid sequential pipelines. By using **LangGraph**, we transform AxiomAlpha into an autonomous, state-driven multi-agent system. This allows the system to:
- **Adapt to failures** (e.g., if risk limits are breached, it can recursively route back to the strategy agent).
- **Execute in parallel** (e.g., ML predictions and NLP sentiment analysis can happen concurrently, saving execution time).
- **Maintain a robust audit trail** of every decision made by every agent at every step of the orchestration.

### 2. What We Did (The Implementation)
We orchestrated 7 specialized AI Agents into a state machine with conditional routing:
1. **DataAgent**: Validates incoming market data and features.
2. **QuantAgent**: Identifies the current market regime (Bull, Bear, Volatile).
3. **GraphAgent**: Analyzes systemic risk and network centrality among assets.
4. **MLAgent**: Generates alpha signals using machine learning models.
5. **NLPAgent**: Parses news and macro reports to gauge market sentiment.
6. **RiskAgent**: Acts as the system's "brakes," auditing the proposed portfolio against strict risk limits.
7. **StrategyAgent**: The final decision-maker that synthesizes all signals into target portfolio weights.

### 3. Key Concepts & Formulas Used
During the orchestration, the agents rely on several quantitative models:
- **Value at Risk (VaR) & Conditional Value at Risk (CVaR)**: Used by the RiskAgent to assess tail risk. $VaR$ answers "what is the most I can expect to lose with 95% confidence?", while $CVaR$ answers "if I do lose more than $VaR$, how bad will it be?".
- **Regime Classification**: Used by the QuantAgent to classify the market state based on volatility ($v$) and momentum ($m$).
- **Mean-Variance Optimization**: Used by the StrategyAgent to maximize expected return for a given level of risk: $\max_w (w^T \mu - \frac{\lambda}{2} w^T \Sigma w)$
- **Network Centrality**: Used by the GraphAgent to penalize assets that are too highly interconnected, thus reducing systemic risk exposure.

### 4. Insights Gained from the Orchestrated Analysis
Running this pipeline gives us a comprehensive, unified view of the market:
- **Dynamic Asset Allocation**: The system shifts cash and asset weights dynamically based on the identified market regime.
- **Risk-Adjusted Alpha**: We don't just chase returns; every signal is rigorously audited by the RiskAgent, ensuring that risk limits are respected.
- **Explainable AI in Finance**: Because each agent contributes to the `GraphState`, we can precisely trace *why* the StrategyAgent chose a specific allocation (e.g., "Cash was raised to 30% because the RiskAgent flagged elevated volatility").

### LangGraph Node Functions

Each agent becomes a node function.
Node functions take state and return state.
LangGraph calls them in graph order.

Key difference from raw agent classes:
  Node functions must be pure functions
  They instantiate agents internally
  They handle all errors and return valid state

```python
# Agent node implementations.
#
# Each node takes the shared `state` dict (which travels through the
# LangGraph state machine), reads the relevant pipeline artifacts via
# `src.io.loaders`, condenses them into a small JSON-serializable summary,
# and merges that summary back into `state`. No LLM calls — explanations
# are templated.
import json as _json
import numpy as _np
import pandas as _pd
from src.io.loaders import (
    load_log_returns, load_macro,
    load_risk_metrics, load_regimes,
    load_centrality_scores, load_systemic_risk_scores,
    load_master_signals, load_sentiment_features,
    load_optimal_weights, load_performance_metrics,
)


def _safe(call, default=None):
    """Call a loader, return `default` if the artifact is missing/malformed."""
    try:
        return call()
    except Exception as exc:
        print(f"   [WARN] {call.__name__ if hasattr(call, '__name__') else call}: "
              f"{type(exc).__name__}: {exc}")
        return default


def node_data(state):
    """Validate every input feed actually loads and meets minimum size."""
    print("\nDataAgent running...")
    log_ret = _safe(load_log_returns)
    macro   = _safe(load_macro)
    report = {
        'log_returns_rows':    int(len(log_ret)) if log_ret is not None else 0,
        'log_returns_tickers': int(log_ret.shape[1]) if log_ret is not None else 0,
        'macro_rows':          int(len(macro))   if macro is not None else 0,
        'first_date':          str(log_ret.index.min().date()) if log_ret is not None else None,
        'last_date':           str(log_ret.index.max().date()) if log_ret is not None else None,
    }
    is_valid = report['log_returns_rows'] >= 252 and report['log_returns_tickers'] >= 5
    state['data_report'] = report
    state['data_valid']  = bool(is_valid)
    if not is_valid:
        state['abort'] = True
        print("  [FAIL] Data quality below threshold — aborting downstream agents.")
    else:
        print(f"  ✓ {report['log_returns_rows']} days × {report['log_returns_tickers']} tickers loaded.")
    return state


def node_quant(state):
    """Summarize current regime + average risk metrics."""
    print("\nQuantAgent running...")
    rm = _safe(load_risk_metrics)
    rg = _safe(load_regimes)
    quant_summary = {}
    if rg is not None and len(rg):
        last = rg.iloc[-1]
        quant_summary['current_regime_encoded'] = (
            int(last['regime_encoded']) if 'regime_encoded' in rg.columns
            and _pd.notna(last['regime_encoded']) else None
        )
        if 'regime_name' in rg.columns:
            quant_summary['current_regime_name'] = str(last['regime_name'])
    if rm is not None and len(rm):
        latest = rm.sort_values('Date').groupby('ticker').tail(1)
        for col in ('garch_vol', 'var_95', 'cvar_95', 'downside_vol'):
            if col in latest.columns:
                quant_summary[f'mean_{col}'] = float(latest[col].mean())
    # Allocation regime is calculated from the optimizer's trailing window.
    # Use that same value here instead of a differently defined regime series.
    try:
        with open(DATA_PROC / 'portfolio_metrics.json') as f:
            _allocation_metadata = _json.load(f)
        _regime = _allocation_metadata.get('regime')
        if _regime in {'Bear', 'Volatile', 'Bull'}:
            quant_summary['current_regime_name'] = _regime
            quant_summary['current_regime_encoded'] = {'Bear': 0, 'Volatile': 1, 'Bull': 2}[_regime]
    except (OSError, ValueError):
        pass
    state['quant_summary'] = quant_summary
    print(f"  ✓ regime={quant_summary.get('current_regime_name', '?')}, "
          f"risk metrics summarized.")
    return state


def node_graph(state):
    """Identify top systemic-risk tickers from the centrality artifact."""
    print("\nGraphAgent running...")
    cent = _safe(load_centrality_scores)
    sys_ = _safe(load_systemic_risk_scores)
    graph_summary = {}
    if sys_ is not None and 'systemic_risk_score' in sys_.columns:
        top = sys_['systemic_risk_score'].sort_values(ascending=False).head(5)
        graph_summary['top_systemic_tickers'] = [
            {'ticker': str(t), 'score': float(v)} for t, v in top.items()
        ]
    if cent is not None and 'pagerank' in cent.columns:
        top_pr = cent['pagerank'].sort_values(ascending=False).head(5)
        graph_summary['top_pagerank_tickers'] = [
            {'ticker': str(t), 'pagerank': float(v)} for t, v in top_pr.items()
        ]
    state['graph_summary'] = graph_summary
    n = len(graph_summary.get('top_systemic_tickers', []))
    print(f"  ✓ {n} systemic-risk anchors identified.")
    return state


def node_ml(state):
    """Pull per-ticker ML direction probabilities (latest available date)."""
    print("\nMLAgent running...")
    sig = _safe(load_master_signals)
    ml_summary = {}
    if sig is not None and len(sig):
        date_col = 'Date' if 'Date' in sig.columns else ('date' if 'date' in sig.columns else None)
        if date_col is not None:
            sig = sig.copy()
            sig[date_col] = _pd.to_datetime(sig[date_col])
            latest_date = sig[date_col].max()
            latest = sig[sig[date_col] == latest_date]
            ml_summary['as_of'] = str(latest_date.date())
            if 'training_end' in latest.columns:
                ml_summary['training_end'] = str(_pd.to_datetime(latest['training_end']).max().date())
            if 'alpha_signal' in latest.columns and 'ticker' in latest.columns:
                top_long = latest.nlargest(5, 'alpha_signal')[['ticker', 'alpha_signal']]
                ml_summary['top_long_signals'] = [
                    {'ticker': str(t), 'p_up': float(p)} for t, p in
                    zip(top_long['ticker'], top_long['alpha_signal'])
                ]
    state['ml_summary'] = ml_summary
    print(f"  ✓ {len(ml_summary.get('top_long_signals', []))} long-side ML signals selected.")
    return state


def node_nlp(state):
    """Aggregate per-ticker sentiment_score over the trailing 30 days."""
    print("\nNLPAgent running...")
    sf = _safe(load_sentiment_features)
    nlp_summary = {'source': 'synthetic; not used for allocation'}
    if sf is not None and len(sf):
        sf = sf.copy()
        sf['Date'] = _pd.to_datetime(sf['Date'])
        cutoff = sf['Date'].max() - _pd.Timedelta(days=30)
        recent = sf[sf['Date'] >= cutoff]
        if 'sentiment_score' in recent.columns:
            avg = recent.groupby('ticker')['sentiment_score'].mean()
            nlp_summary['avg_sentiment_30d'] = {
                str(t): float(v) for t, v in avg.items() if _pd.notna(v)
            }
            nlp_summary['most_positive'] = avg.idxmax() if len(avg) else None
            nlp_summary['most_negative'] = avg.idxmin() if len(avg) else None
    state['nlp_summary'] = nlp_summary
    print(f"  ✓ sentiment aggregated over {len(nlp_summary.get('avg_sentiment_30d', {}))} tickers.")
    return state


def node_risk(state):
    """Audit overall portfolio risk against the latest backtest scorecard."""
    print("\nRiskAgent running...")
    perf = _safe(load_performance_metrics)
    risk_audit = {}
    if isinstance(perf, dict) and perf:
        risk_audit['perf_dict'] = {
            str(k): {str(kk): float(vv) if isinstance(vv, (int, float)) else str(vv)
                     for kk, vv in (v.items() if isinstance(v, dict) else {'value': v}.items())}
            for k, v in perf.items()
        }
        # Simple guard rail: if AxiomAlpha max_dd worse than -50%, flag.
        aa = perf.get('AxiomAlpha') if isinstance(perf.get('AxiomAlpha'), dict) else None
        if aa and 'max_dd' in aa and aa['max_dd'] < -0.50:
            risk_audit['alert'] = f"AxiomAlpha max_dd {aa['max_dd']:.1%} exceeds -50% guardrail"
            state['risk_alert'] = True
    if (not isinstance(perf, dict) or not isinstance(perf.get('AxiomAlpha'), dict)
            or 'max_dd' not in perf.get('AxiomAlpha', {})
            or not _np.isfinite(perf['AxiomAlpha']['max_dd'])):
        risk_audit['alert'] = 'AxiomAlpha backtest risk metrics are unavailable'
        state['risk_alert'] = True
    # Validate the allocation itself in addition to the historical drawdown.
    from src.portfolio import audit_allocation
    candidate = _safe(load_optimal_weights)
    if candidate is None or not {'Date', 'ticker', 'weight'}.issubset(candidate.columns):
        allocation_errors = ['allocation is unavailable']
    else:
        latest = candidate[candidate['Date'] == candidate['Date'].max()]
        allocation_errors = audit_allocation(latest.set_index('ticker')['weight'])
        last_data_date = state.get('data_report', {}).get('last_date')
        if last_data_date is not None and _pd.Timestamp(latest['Date'].max()).normalize() != _pd.Timestamp(last_data_date).normalize():
            allocation_errors.append('allocation date does not match the latest market data')
    risk_audit['allocation_errors'] = allocation_errors
    if allocation_errors:
        risk_audit['alert'] = '; '.join(allocation_errors)
        state['risk_alert'] = True
    state['risk_audit'] = risk_audit
    print(f"  ✓ risk audit complete (alert={'risk_alert' in state})")
    return state


def node_strategy(state):
    """Read optimal_weights and assemble the final decision payload."""
    print("\nStrategyAgent running...")
    if state.get('risk_alert', False):
        state['strategy_decision'] = {'status': 'blocked', 'rationale': state.get('risk_audit', {}).get('alert', 'Risk check failed')}
        return state
    ow = _safe(load_optimal_weights)
    decision = {}
    if ow is not None and len(ow):
        latest_date = ow['Date'].max() if 'Date' in ow.columns else None
        latest = ow[ow['Date'] == latest_date] if latest_date is not None else ow
        if 'ticker' in latest.columns and 'weight' in latest.columns:
            decision['as_of'] = str(_pd.to_datetime(latest_date).date()) if latest_date is not None else None
            top = latest.nlargest(10, 'weight')[['ticker', 'weight']]
            decision['top_holdings'] = [
                {'ticker': str(t), 'weight': float(w)} for t, w in
                zip(top['ticker'], top['weight'])
            ]
            decision['cash_pct'] = float(max(0.0, 1.0 - latest['weight'].sum()))
    decision['rationale'] = (
        "Templated explanation: position sized by AxiomAlpha optimizer "
        "(see NB11). Regime overlay and systemic-risk penalty are baked "
        "into the weights. Risk audit: "
        f"{state.get('risk_audit', {}).get('alert', 'no guardrail breaches')}."
    )
    state['strategy_decision'] = decision
    print(f"  ✓ strategy decision assembled "
          f"(top_holdings={len(decision.get('top_holdings', []))}, "
          f"cash={decision.get('cash_pct', 0):.1%}).")
    return state


def dict_to_agent_state(d: dict):
    """Compatibility shim — the new node bodies operate on dict directly."""
    return d


def agent_state_to_dict(state) -> dict:
    return state if isinstance(state, dict) else dict(state)



# --- LangGraph routing predicates ----------------------------------------

def should_abort(state):
    """Edge predicate after node_data: abort if data invalid, else continue."""
    return 'abort' if state.get('abort', False) else 'continue'


def check_risk(state):
    """Edge predicate after node_risk: re-route to strategy in either case
    (the 'retry' branch exists for future expansion; today both go forward)."""
    return 'block' if state.get('risk_alert', False) else 'finalize'
```

### Building the LangGraph

Now we wire all nodes together with edges.
The graph defines:
  Which agents run in which order
  What conditions trigger which paths
  Where the pipeline ends

**Visualization 1: AxiomAlpha Agent Graph**
The following code generates a networkx diagram illustrating the state machine flow, conditional routes, and terminal nodes of the agent orchestration framework.

```python
if LANGGRAPH_AVAILABLE:
    # Build graph
    workflow = StateGraph(dict)
    
    # Add nodes
    workflow.add_node("data", node_data)
    workflow.add_node("quant", node_quant)
    workflow.add_node("graph", node_graph)
    workflow.add_node("ml", node_ml)
    workflow.add_node("nlp", node_nlp)
    workflow.add_node("risk", node_risk)
    workflow.add_node("strategy", node_strategy)
    
    # Entry point
    workflow.set_entry_point("data")
    
    # Conditional edge after DataAgent
    workflow.add_conditional_edges(
        "data",
        should_abort,
        {
            'abort': END,
            'continue': 'quant'
        }
    )
    
    # After QuantAgent: run Graph, ML, NLP
    # (in sequence — parallel requires special setup)
    workflow.add_edge("quant", "graph")
    workflow.add_edge("graph", "ml")
    workflow.add_edge("ml", "nlp")
    workflow.add_edge("nlp", "risk")
    
    # After RiskAgent: conditional strategy
    workflow.add_conditional_edges(
        "risk",
        check_risk,
        {
            'block': END,
            'finalize': 'strategy'
        }
    )
    
    # Strategy → END
    workflow.add_edge("strategy", END)
    
    # Compile
    app = workflow.compile()
    print("LangGraph compiled successfully")
    print(f"Nodes: {list(workflow.nodes.keys())}")

else:
    # Manual sequential pipeline fallback
    print("Using sequential pipeline (no LangGraph)")
    
    def run_pipeline(initial_state):
        state = initial_state
        
        print("\n" + "="*50)
        print("AxiomAlpha Pipeline Starting")
        print("="*50)
        
        state = node_data(state)
        if state.get('abort', False):
            print("Pipeline aborted: data quality")
            return state
        
        state = node_quant(state)
        state = node_graph(state)
        state = node_ml(state)
        state = node_nlp(state)
        state = node_risk(state)
        if not state.get('risk_alert', False):
            state = node_strategy(state)
        
        print("\n" + "="*50)
        print("AxiomAlpha Pipeline Complete")
        print("="*50)
        return state

print("Pipeline ready to run")

```
